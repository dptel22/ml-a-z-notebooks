Exercise 1. Arithmetic Product and Conditional Logic

```python
num1 = int(input("Num1: "))
num2 = int(input("Num2: "))

if num1 * num2 <= 1000:
    product = num1 * num2
    print(product)
else:
    total = num1 + num2
    print(total)
```

Exercise 2. Cumulative Sum of a Range

```python
print("Printing current and previous number sum in a range(10)")
previous_num = 0

for i in range(10):
  x_sum = previous_num + i
  print(f"Current Number {i} Previous Number {previous_num} Sum: {x_sum}")

  previous_num = i
```

Exercise 9. Vowel Frequency Counter

```python
vowels = "aeiou"
sentence = "Learning Python is fun!"
count = 0

for i in sentence.lower():
  if i in vowels:
    count += 1

print(count)
```

Exercise 12. List Comparison and Boolean Logic

```python
def first_last_same(number_list):
  first_num = number_list[0]
  last_num = number_list[-1]

  if first_num == last_num:
    return True
  else:
    return False

numbers_x = [10, 20, 30, 40, 10]
numbers_y = [75, 65, 35, 75, 30]

print("result for x: ", first_last_same(numbers_x))
print("result for y: ", first_last_same(numbers_y))
```

Exercise 16. Numerical Palindrome Check

```python
number = list(input("Enter a number: "))


reverse = number[::-1]

if reverse == number:
   print("True")

else:
  print("False")
```
