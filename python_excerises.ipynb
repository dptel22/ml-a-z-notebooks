Exercise 12. List Comparison and Boolean Logic

```python
def first_last_same(number_list):
  first_num = number_list[0]
  last_num = number_list[-1]

  if first_num == last_num:
    return True
  else:
    return False

numbers_x = [10, 20, 30, 40, 10]
numbers_y = [75, 65, 35, 75, 30]

print("result for x: ", first_last_same(numbers_x))
print("result for y: ", first_last_same(numbers_y))
```

```python
income = 45000
tax = 0

if income <= 10000:
  tax = 0

elif income <= 20000:
  tax = (income - 10000) * 10 / 100

else:
  tax = 0 + (10000 * 10/100)
  tax += (income - 20000) * 20/100

print(tax)
```

Exercise 31. Print Alternate Prime Numbers

```python
def is_prime(num):
  if num < 2:
    return False

  for i  in range(2, int(num ** 0.5) + 1):
    if num % i == 0:
      return False
  return True

limit = 20
counter = 0

for n in range(2, limit +1):
  if is_prime(n):
    if counter % 2 == 0:
      print(n, end= " ")
    counter +=1
```

Exercise 40. Introduction to Classes (OOP)

```python
class Car:
  def __init__(self, make, model, year):
    self.make = make
    self.model = model
    self.year = year

  def start_engine(self):
    print(f"The {self.year} {self.make} {self.model} engine is now running")


my_car = Car("Toyota", "ISIS", 2015)

my_car.start_engine()
```
