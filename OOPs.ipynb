Exercise 2: Vehicle Class with Instance Attributes

```python
class Vehicle:
  def __init__(self, name, max_speed, mileage):
    self.name = name
    self.max_speed = max_speed
    self.mileage = mileage

vechile = Vehicle("Tesla Model S", 250, 18)
print(vechile.name, vechile.max_speed, vechile.mileage)
```

Exercise 3: Rectangle Class with Area & Perimeter

```python
class Rectangle:
  def __init__ (self, length, width):
    self.length = length
    self.width = width

  def perimeter(self):
    return 2 * (self.length + self.width)

  def Area(self):
    return self.length * self.width

rect = Rectangle(10, 4)
print(rect.perimeter(), rect.Area())
```

Exercise 4: Student Class with Average Grade

```python
class Student:
  def __init__(self, name, marks):
    self.name = name
    self.marks = marks

  def average(self):
    average = sum(self.marks) / len(self.marks)
    return average


avg = Student("Alice", [85, 90, 78, 92, 88])
print(f"{avg.name} average marks is: {avg.average()}")
```

Exercise 5: Product Class with Stock Value Calculator


```python
class Product:
  def __init__(self, name, price, quantity):
    self.name = name
    self.price = price
    self.quantity = quantity

  def total_value(self):
    return self.price * self.quantity


p1 = Product("Laptop", 899.99, 5)
print(f"{p1.name} has stock value of {p1.total_value()}")
```
