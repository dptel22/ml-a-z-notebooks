Exercise 22: Remove common items from array

```python
a = np.array([1, 2, 3, 4, 5])
b = np.array([5, 6, 7, 8, 9])

mask = np.in1d(a, b)
result = a[~mask]
print(result)
```

Exercise 23: Normalize a NumPy array (values between 0 and 1)

```python
arr = np.array([10, 20, 30, 40, 50])

arr = (arr - arr.min()) / (arr.max() - arr.min())
print(arr)
```

Exercise 27: Sorting a NumPy array based on a specific column

```python
sampleArray = np.array([[34, 43, 73], [82, 22, 12], [53, 94, 66]])
index = sampleArray[:, 1].argsort()
sorted_array = sampleArray[index]
print(sorted_array)

```

Exercise 28: Delete and Insert a Column in a NumPy Array

```python
sampleArray = np.array([[34,43,73],[82,22,12],[53,94,66]])
newColumn = np.array([10, 10, 10])

deleteArray = np.delete(sampleArray, 1, axis=1)
sampleArray = np.insert(deleteArray, 1, newColumn, axis=1)
print(sampleArray)
```
